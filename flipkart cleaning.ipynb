```python
import pandas as pd
df = pd.read_csv("flipkart.csv")
df
```

```python
df.shape
```

```python
df.info()
```

```python
df = df.copy()
df.dropna(axis=0, inplace=True)
df
```

```python
df.info()
```

```python
df.replace({
    'rating' : ('[Ratings]','[,]'),
    'price' : ('[₹]','[,]')
},'',regex=True,inplace=True)
df
```

```python
df['brand'] = df['brand'].replace(r'\s*\(.*?\)\s*', '', regex=True)
df
```

```python
df.price=df.price.astype(int)
df.price.dtype
```

```python
df.rating=df.rating.astype(int)
df.rating.dtype
```
