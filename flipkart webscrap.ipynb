```python
import requests
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
```

```python
driver = webdriver.Chrome()
driver.get("https://www.flipkart.com/")
```

```python
search = driver.find_element(By.XPATH, '/html/body/div[1]/div/div[1]/div/div/div/div/div[1]/div/div/div/div[1]/div[1]/header/div[1]/div[2]/form/div/div/input')
search.send_keys("Mobiles")
search.send_keys(Keys.ENTER)
```

```python
import time
links=[]
data=driver.find_elements(By.CLASS_NAME,'CGtC98')
for i in data:
     links.append(i.get_attribute("href"))
button=driver.find_element(By.XPATH,'//*[@class="_9QVEpD"]')
button.click()
time.sleep(1)
for i in range(15):
    data=driver.find_elements(By.CLASS_NAME,'CGtC98')
    for i in data:
        links.append(i.get_attribute("href"))
    button=driver.find_element(By.XPATH,'//*[@class="_9QVEpD"][2]')
    button.click()
    time.sleep(1)
print(links[0:3])
```

```python
prices = []
discounts = []
ratings = []
stars = []
brands = []
storage = []
processor = []
```

```python
j=6
for i in links[370:]:   #-- we can start the 'ith' value from 1.--
    driver.get(i)
    WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.XPATH, '//*[@class="hl05eU"]')))
    try:
        prices.append(driver.find_element(By.XPATH, '//*[@class="hl05eU"]//*[@class="Nx9bqj CxhGGd"]').text)
    except:
        prices.append(None)
    try:
        discounts.append(driver.find_element(By.XPATH, '//*[@class="UkUFwK WW8yVX"]//span').text)
    except:
        discounts.append(None)
    try:
        ratings.append(driver.find_element(By.XPATH, '//*[@class="Wphh3N"]//span//span').text)
    except:
        ratings.append(None)
    try:
        stars.append(driver.find_element(By.XPATH, '//*[@class="XQDdHH"]').text)
    except:
        stars.append(None)
        
    try:
        brands.append(driver.find_element(By.XPATH, '//*[@class= "C7fEHH"]//div//h1//span').text)
    except:
        brands.append(None)

    time.sleep(5)
    button=driver.find_element(By.XPATH,'//*[@class= "_3Fm-hO"]//div//button')
    button.click()
    
    try:
        processor.append(driver.find_element(By.XPATH, '//*[@class="_1OjC5I"]//div[3]//*[@class= "_0ZhAN9"]//tbody//tr[2]//td//ul//li').text)
    except:
        processor.append(None)
    try:
        storage.append(driver.find_element(By.XPATH, '//*[@class= "_3Fm-hO"]//div[4]//*[@class= "_0ZhAN9"]//tr[1]//td//ul//li').text)
    except:
        storage.append(None)
    print(j)
    j=j+1
```

```python
print("stars :",len(stars),"pri :",len(prices),"ratings :",len(ratings),"disc :",len(discounts),"brand :",len(brands),"stg :",len(storage),"processor :",len(processor))  # -- below values are total data scrap values
```

```python
df=pd.DataFrame({"star":stars,"rating":ratings,"discount":discounts,"price":prices,"brand":brands,"storage":storage,"processor":processor})
df.head()
```
